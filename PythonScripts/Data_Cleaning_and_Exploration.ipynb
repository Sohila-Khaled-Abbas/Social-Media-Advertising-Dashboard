```python
# Import Necessary Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the Dataset
df = pd.read_csv('Social_Media_Advertising.csv')
```

# Inspect the Data

```python
# View the first few rows:
print(df.head())
```

```python
# Check the data types:
print(df.dtypes)
```

```python
# Check for missing values:
print(df.isnull().sum())
```

# Data Type Conversion

```python
# Remove dollar sign and convert Acquisition_Cost to float:
df['Acquisition_Cost'] = df['Acquisition_Cost'].replace(r'[\$,]', '', regex=True).astype(float)
```

```python
# Convert Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'])
```

```python
# Convert ROI to numeric:
df['ROI'] = df['ROI'].astype(float)
```

# Create New Features

```python
# Calculate Total_Spend:
df['Total_Spend'] = df['Acquisition_Cost'] * df['Clicks']
```

# Preliminary Data Visualization in Python

```python
# Distribution of ROI:
sns.histplot(df['ROI'], kde=True)
plt.title('Distribution of ROI')
plt.xlabel('ROI')
plt.ylabel('Frequency')
plt.show()
```

```python
# Number of campaigns per company:
company_counts = df['Company'].value_counts()
company_counts.plot(kind='bar')
plt.title('Number of Campaigns per Company')
plt.xlabel('Company')
plt.ylabel('Number of Campaigns')
plt.show()
```

```python
# Conversion Rate vs ROI:
sns.scatterplot(x='Conversion_Rate', y='ROI', data=df)
plt.title('Conversion Rate vs ROI')
plt.xlabel('Conversion Rate')
plt.ylabel('ROI')
plt.show()
```

# Save the Cleaned Data

```python
df.to_csv('Social_Media_Advertising_cleaned.csv', index=False)
```
